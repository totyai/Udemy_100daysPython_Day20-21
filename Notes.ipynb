*Inheritence & List Slicing*

To inherit from a calls in child class's init should look like this:
```
class Fish(Animal):
    def __init__(self):
        super().__init__()
```
Where the *Animal* is the _Parent_ class, and the *super* also refers to the _Parent_'s init method.

This enables the *Fish* child class to use the attributes or methods of the *Animal* class.

````python
#If the two classes are not connected
class Animal:
    def __init__(self):
        self.num_eyes = 2
    
    def breathe(self):
        print("Inhale, exhale.")

class Fish:

    def swim(self):
        print("moving in water.")
    
nemo = Fish()
nemo.swim()
nemo.breathe() # - should get an error
````

````python
#If the two classes are  connected
class Animal:
    def __init__(self):
        self.num_eyes = 2
    
    def breathe(self):
        print("Inhale, exhale.")

class Fish(Animal):
    def __init__(self):
        super().__init__()

    def swim(self):
        print("moving in water.")
    
nemo = Fish()
nemo.swim()
nemo.breathe() #is available since of the connection
print(nemo.num_eyes)
````

We can also further modify the parent's class in the child.
Like the *fish* cannot breathe, but add also _doing this underwater_ as an extra.

````python
#Modify parent's method
class Animal:
    def __init__(self):
        self.num_eyes = 2
    
    def breathe(self):
        print("Inhale, exhale.")

class Fish(Animal):
    def __init__(self):
        super().__init__()

    def breathe(self):
        super().breathe() #So it is doing the same as the Animal
        print("doing this underwater.") #But also this as extra

    def swim(self):
        print("moving in water.")
    
nemo = Fish()
nemo.breathe()
````
